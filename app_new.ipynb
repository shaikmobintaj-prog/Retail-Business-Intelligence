```python
df=df.dropna()
```

```python
df["Sales"] = df["Quantity"] * df["Price"]

df["Profit"] = df["Quantity"] * (df["Price"] - df["Cost"])

df["Profit_Margin"] = (df["Profit"] / df["Sales"] * 100)
```

```python
def welcome():
    print(" Welcome to Retail BI Analysis")
```

```python
def calculate_average_sales():
    return df["Sales"].mean()


welcome()

print("Average Sales:", calculate_average_sales())
```

```python
class RetailBusiness():
    def __init__(self, data):
        self.data = data
    def total_sales(self): 
        return self.data["Sales"].sum()
    def total_profit(self):
        return self.data["Profit"].sum()

    def total_orders(self):
        return len(self.data)

    def total_quantity(self):
        return self.data["Quantity"].sum()

    def average_sales(self):
        return self.data["Sales"].mean()

    def best_product(self):
        return self.data.groupby("Product")["Sales"].sum().idxmax()

    def best_category(self):
        return self.data.groupby("Category")["Sales"].sum().idxmax()

    def best_region(self):
        return self.data.groupby("Region")["Sales"].sum().idxmax()

    def product_report(self):
        return self.data.groupby("Product")["Sales"].sum().sort_values(ascending=False)

    def category_report(self):
        return self.data.groupby("Category")["Sales"].sum()

    def region_report(self):
        return self.data.groupby("Region")["Sales"].sum()

    def profit_report(self):
        return self.data.groupby("Product")["Profit"].sum().sort_values(ascending=False)

    
```

```python
retail = RetailBusiness(df)
(retail.total_sales(),
retail.total_profit(),
retail.total_orders(),
retail.total_quantity(),
retail.average_sales(),
retail.best_product(),
retail.best_category(),
retail.best_region())
```

```python
Product_Sales = df.groupby('Product')['Sales'].sum()

Product_Sales.plot(kind="bar", color="skyblue")

plt.title("Product Wise Sales")

plt.xlabel("Product")

plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
Category_Sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(7, 7))

plt.pie(Category_Sales,labels=Category_Sales.index,autopct="%1.1f%%",startangle=90)

plt.title("Category Sales Distribution")

plt.show()

```

```python
Region_Sales = df.groupby('Region')['Sales'].sum()


plt.figure(figsize=(8, 5))

sns.barplot(x=Region_Sales.index,y=Region_Sales.values)

plt.title("Region Wise Sales")

plt.xlabel("Region")

plt.ylabel("Sales")

plt.show()
```
